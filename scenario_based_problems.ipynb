```python
# Step 1: Import required libraries
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Step 2: Load the dataset
df = pd.read_csv("diabetes_prediction_dataset.csv")

# Display 
print("Shape:", df.shape)
print(df.head())

# Step 3: Identify the data type of each feature
print("\nData Types:")
print(df.dtypes)

# Step 4: Separate numerical and categorical attributes
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns
categorical_cols = df.select_dtypes(include=['object']).columns

print("\nNumerical Columns:")
print(list(numerical_cols))

print("\nCategorical Columns:")
print(list(categorical_cols))
# Step 5: Convert categorical variables into numerical form
label_encoder = LabelEncoder()

for col in categorical_cols:
    df[col] = label_encoder.fit_transform(df[col])

print("\nDataset after Encoding:")
print(df.head())
# Step 6: Remove unnecessary columns (if any)
# Example: Remove 'id' column if it exists
if 'id' in df.columns:
    df.drop('id', axis=1, inplace=True)
    print("\n'id' column removed.")
else:
    print("\nNo unnecessary 'id' column found.")

# Step 7: Display summary statistics
print("\nSummary Statistics:")
print(df.describe())

# Display dataset information
print("\nDataset Information:")
print(df.info())

# Step 8: Save the cleaned dataset
df.to_csv("diabetes_prediction_dataset_cleaned.csv", index=False)

print("\nCleaned dataset saved as 'diabetes_prediction_dataset_cleaned.csv'")
```

```python
# ============================================
# Experiment 2: Handling Missing Values
# Dataset: Kaggle Titanic (train.csv)
# ============================================

# Import required libraries
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("train.csv")

# --------------------------------------------
# Display dataset before preprocessing
# --------------------------------------------
print("===== DATASET BEFORE PREPROCESSING =====\n")
print(df.head())

print("\nDataset Information:")
print(df.info())

# --------------------------------------------
# Detect missing values
# --------------------------------------------
print("\n===== MISSING VALUES (True = Missing) =====\n")
print(df.isnull())

# --------------------------------------------
# Count missing values in each column
# --------------------------------------------
print("\n===== MISSING VALUES COUNT =====\n")
print(df.isnull().sum())

# --------------------------------------------
# Replace numerical missing values using Mean
# --------------------------------------------
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

for col in numeric_columns:
    if df[col].isnull().sum() > 0:
        mean_value = df[col].mean()
        df[col].fillna(mean_value, inplace=True)

# --------------------------------------------
# Replace categorical missing values using Mode
# --------------------------------------------
categorical_columns = df.select_dtypes(include=["object"]).columns

for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()[0]
        df[col].fillna(mode_value, inplace=True)

# --------------------------------------------
# Display dataset after preprocessing
# --------------------------------------------
print("\n===== DATASET AFTER PREPROCESSING =====\n")
print(df.head())

print("\nUpdated Dataset Information:")
print(df.info())

# --------------------------------------------
# Verify missing values after preprocessing
# --------------------------------------------
print("\n===== MISSING VALUES COUNT AFTER CLEANING =====\n")
print(df.isnull().sum())

# --------------------------------------------
# Compare before and after
# --------------------------------------------
print("\n===== COMPARISON =====")
print("Shape of Dataset:", df.shape)

print("\nMissing Values Before Cleaning:")
before = pd.read_csv("train.csv").isnull().sum()
print(before)

print("\nMissing Values After Cleaning:")
after = df.isnull().sum()
print(after)
```
